```python
#W7A1 Viết hàm tìm kiếm nhị phân trên một danh sách đã
#được sắp xếp tăng dần. Trả về chỉ số của phần tử hoặc -1 nếu không tìm
#thấy. mẫu hàm: def binary_search(arr, left, right, target):

def binary_search(arr, left, right, target):
    while left <= right:
        mid = (left + right) // 2

        if arr[mid] == target:
            return mid
        elif arr[mid] < target:
            left = mid + 1
        else:
            right = mid - 1

    return -1

arr = list(map(int, input("Nhập các phần tử (đã sắp xếp tăng dần): ").split()))
target = int(input("Nhập giá trị cần tìm: "))

result = binary_search(arr, 0, len(arr) - 1, target)

print(result)



```

```python
#W7A2 Cho danh sách arr và một
#số nguyên x, đếm xem x
#xuất hiện bao nhiêu lần
#trong danh sách. def
#count_occurrences(arr, x) -> int:

def count_occurrences(arr,x) -> int:
    cnt = 0
    for i in arr:
        if i == x:
            cnt += 1
    return cnt

n = list(map(int, input().split()))
x = int(input())
print(count_occurrences(n,x))
```

```python
#W7A3
# Viết hàm thực hiện thuật
# toán bubble sort, Selection
# Sort để sắp xếp danh sách
# theo thứ tự tăng dần. Sắp
# xếp nổi bọt (Bubble Sort); Sắp xếp chọn (Selection
# Sort)

def bubble_sort(lst):
    n = len(arr)
    for i in range(n-1):
        for j in range(0,n-1-i):
            if lst[j] > lst[j+1]:
                lst[j], lst[j+1] = lst[j+1], lst[j]
    return lst

n = list(map(int, input().split()))
print(bubble_sort(n))

def selection_sort(lst):
    n = len(lst)
    for i in range(n-1):
        min_idx = i
        for j in range(i+1,n):
            if lst[j] < lst[min_idx]:
                min_idx = j
        lst[i], lst[min_idx] = lst[min_idx], lst[i]
    return lst

n = list(map(int, input().split()))
print(selection_sort(n))
```

```python
#W7A4
# Viết hàm tìm phần tử có số lần xuất hiện nhiều nhất
# trong danh sách. Nếu có nhiều phần tử cùng số lần,
# trả về phần tử xuất hiện trước.

def appear_time(num,lst):
    cnt = 0
    for i in lst:
        if i == num:
            cnt += 1
    return cnt
def most_appear(lst):
    max_appear = 0
    min_num = 0 
    for num in lst:
        if appear_time(num,lst) > max_appear or (appear_time(num,lst) == max_appear and num < min_num):
            max_appear = appear_time(num,lst)
            min_num = num
    return f'So xuat hien nhieu nhat: {min_num}, so lan xuat hien: {max_appear}'
n = list(map(int, input().split()))
print(most_appear(n))
```

```python
#W7A5
# Cho danh sách các số nguyên và một số X, đếm
# tất cả các cặp (a, b) sao cho a + b = X. c

line = list(map(int, input().split()))
x = int(input())

def count_pairs(arr,x):
    dic = {}
    cnt = 0
    for a in arr:
        b = x - a
        if b in dic:
            cnt += dic[b]
        dic[a] = dic.get(a,0) + 1
    return cnt

print(count_pairs(line,x))
```

```python
#W7A6
# Bạn được cung cấp một danh sách các danh sách
# lồng nhau, vd [[1, 2, 3], [4, 5], [6, 7, 8, 9]], hãy làm
# phẳng danh sách

def flat_list(arr):
    result = []
    for lst in arr:
        for num in lst:
            result.append(num)
    return result


print(flat_list([[1, 2, 3], [4, 5], [6, 7, 8, 9]]))
```

```python
#W7A7
# Cho một danh sách các số nguyên nums, hãy tìm độ
# dài của dãy con tăng nghiêm ngặt dài nhất
# (không nhất thiết phải liền kề).

line = list(input().split()) #[10, 9, 2, 5, 3, 7, 101, 18]

def len_strictly_increase(arr):
    n = len(arr) 
    stor = [1] * n
    for i in range(n):
        for j in range(i):
            if arr[j] < arr[i]:
                stor[i] = max(stor[i], stor[j]+1)
    return max(stor)

print(len_strictly_increase(line))
```

```python
#W7A8
# Với mỗi truy vấn q, hãy tìm độ dài nhỏ nhất của
# khoảng thời gian [li, ri] sao cho li ≤ q ≤ ri.
# Nếu không tồn tại khoảng thời gian nào như vậy, hãy
# trả về -1 cho truy vấn đó. 
# solve in < O(m.n), hint: binary search in the queries list

def check(arr,lst):
    res = []
    for i in range (len(lst)):
        temp = []
        for j in range (len(arr)):
            if lst[i] >= arr[j][0] and lst[i] <= arr[j][1]:
                temp.append(arr[j][1] - arr[j][0] + 1)
        if len(temp) == 0:
            res.append(-1)
        else :
            res.append(min(temp))
    return res
arr = [[1,3],[5,7],[9,12]]
lst = [2,4,5,8,10]
print(check(arr,lst))
```

```python
#W7A9
# Cho một danh sách các số nums, với mỗi phần tử nums[i], hãy đếm xem có
# bao nhiêu phần tử bên phải nhỏ hơn nums[i].
# Trả về danh sách các số đếm.
# Efficient solution: using heap or binary search tree

def check(arr):
    res = []
    for i in range (len(arr)):
        count = 0
        for j in range (i,len(arr)):
            if arr[i] > arr[j]:
                count += 1
        res.append(count)
    return res
arr = list(map(int,input().split()))
print(check(arr))
```

```python
#W7A10
# Cho một danh sách các số nguyên nums và một số
# nguyên m, hãy tìm tổng mảng con lớn nhất theo modulo m.
# Tức là, với bất kỳ mảng
# con nums[i:j] nào, hãy tính
# sum(nums[i:j]) % m và trả
# về kết quả lớn nhất có thể.

nums = [7, 11, 14, 2, 8, 4] 
m = 12
def largest_subarray_sum(nums,m):
    n = len(nums)
    res = []
    for i in range(n):
        for j in range(n):
            res.append(sum(nums[j:i]) % m)
    return max(res)

print(largest_subarray_sum(nums,m))
```

```python
#W7A11
# Hãy viết một chương trình để xác định xem có bao nhiêu từ xuất hiện trong một câu.
# Đầu vào: Một chuỗi kí tự trên một dòng. Từ khóa cần tìm ở dòng tiếp theo.

def count_appearance(sent, key_word):
    n = len(key_word)
    cnt = 0
    for i in range(len(sent)):
        if sent[i:i+n] == key_word:
            cnt += 1
    return cnt
sent = "Everyone's watching through your eyes. There's only two options. Win or die, win or die. Win or die, die, die, die, die"
key_word = "die"
print(count_appearance(sent,key_word))

```

```python
#W7A13
# Viết hàm trả về độ dài của dãy hài hòa con dài nhất có
# thể có trong mảng giá trị nguyên \( arr \) có \( n \) phần tử 

def harmony_sequence_len(seq):
    cnt = {}
    for num in seq:
        cnt[num] = cnt.get(num,0)+1
    
    max_len = 0
    for s in cnt:
        if s+1 in cnt:
            current_len = cnt[s] + cnt[s+1]
            max_len = max(max_len, current_len)
    return max_len

nums = list(map(int, input().split()))
print(harmony_sequence_len(nums))
```

```python
#W7A14
# Hãy viết một chương trình tìm kiếm tất cả các số 7 có trong một dãy số.
# Đầu vào: số nguyên n trên một dòng và n số nguyên ở dòng kế tiếp. 

def finding_seven(line):
    res = []
    for idx, i in enumerate(line):
        if i == 7:
            res.append(idx)
    if not res:
        return"Not found"
    return res

n = int(input())
while True:
    line = list(map(int, input().split()))
    if len(line) == n:
        break
print(finding_seven(line))
```

```python
#W7A15
# Hãy giúp Nemo tìm tên của các nhân vật này, biết
# rằng tên trong danh sách khách mời được sắp xếp
# theo vị trí ngồi. 

def finding_nemo(names):
    for i in range(len(names)):
        if names[i] == "Nemo":
            if i == len(names) - 1:
                return f'{names[i-1]} and {names[0]}'
            else:
                return f'{names[i-1]} and {names[i+1]}'

n = int(input())
names = []
for _ in range(n):
    name = input()
    names.append(name)
print(finding_nemo(names))
```
