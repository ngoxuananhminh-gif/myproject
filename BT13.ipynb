```python
#W11A1
#Bất cứ một số nguyên dương nào cũng tồn tại một số nguyên dương là số đảo ngược của nó. 
#Hãy viết hàm def reverse (path: str) -> int: để tìm số nguyên đảo ngược của số n 
#Đầu vào 
#Đầu vào của hàm là một xâu biểu diễn đường dẫn đến tệp. Nội dung tệp chứa duy nhất một số nguyên dương. 
#Đầu ra 
#Hàm trả về số nguyên dương đảo ngược của số trong tệp đầu vào.

def reverse (path: str) -> int:
    file = None
    try:
        file = open(path, 'r', encoding = 'utf-8')
        content = file.read().strip()
        if not content:
            return "File rong"
        return int(content[::-1])

    except FileNotFoundError:
        return "Khong tim thay file"
    except ValueError:
        return "Nhap khong hop le"
    finally:
        if file is None:
            file.close()

result = reverse('minh.txt')
print(result)
    

    
```

```python
#W11A2
#Bạn đang chơi trò chơi đánh cắp với một người bạn, luật chơi như sau: 
#Có một đống đá trên bàn, trong mỗi lượt chơi, các bạn có thể loại bỏ từ 1 đến 3 hòn đá. Người loại bỏ hòn đá cuối cùng sẽ là người chiến thắng. 
#Bạn sẽ là người được ưu tiên là người chơi trước. Cả hai bạn đều rất thông minh và có chiến lược tối ưu cho trò chơi. 
#Viết hàm def canWinNim (path: str) -> bool để xác định xem bạn có thể giành chiến thắng trong trò chơi với số lượng đá đã biết trước không. 
#Gợi ý: nếu trong đống đá có 4 hòn đá, thì bạn sẽ không bao giờ thắng được trò chơi: dù bạn lấy 1,2 hay 3 hòn đá, thì bạn của bạn sẽ luôn là người chiến thắng. 
#Đầu vào 
#Đầu vào của hàm là một xâu biểu diễn đường dẫn đến tệp. Nội dung tệp lưu duy nhất một số nguyên là số lượng đá của trò chơi. 
#Đầu ra 
#Hàm trả về true nếu bạn có thể thắng, ngược lại trả về false.

def canWinNim(path: str) -> bool:
    try:
        with open(path, 'r', encoding='utf-8') as file:
            content = file.read().strip()
            n = int(content)
        return n % 4 != 0

    except FileNotFoundError:
        return False
    except ValueError:
        return False

print(canWinNim('minh.txt'))
```

```python
#W11A3
#Tên của các phù thủy của từng nhà được lưu trong một tệp văn bản. 
#Bạn hãy viết hàm def housesOfHogwarts(path: str) 
#để in ra tên của các phù thủy trong từng nhà (mỗi tên nằm trên một dòng), 
#hàm nhận đầu vào là một xâu kí tự chứa đường dẫn đến tệp lưu trữ.
#Tệp lưu trữ có dòng đầu là một số nguyên n thể hiện số lượng tên có trong tệp và n dòng tiếp theo là các tên của phù thủy.
#Đầu vào
#Đầu vào của hàm là một xâu biểu diễn đường dẫn đến tệp chứa tên các phù thủy trong từng nhà.
#Nội dung mỗi tệp gồm n+1 dòng:
#Dòng đầu tiên chứa số nguyên n;
#n dòng tiếp theo là tên các phù thủy.
#Đầu ra
#In ra màn hình tên tất cả các phù thủy, mỗi tên trên một dòng.

def housesOfHogwarts(path: str):
    file = None
    result = ""

    try:
        file = open(path, 'r', encoding='utf-8')

        first_line = file.readline().strip()
        n = int(first_line)

        for i in range(n):
            name = file.readline().strip()
            result += name + "\n"

    except FileNotFoundError:
        result = "Khong tim thay file"
    except ValueError:
        result = "Du lieu khong hop le"
    finally:
        if file is not None:
            file.close()

    return result.rstrip("\n")


print(housesOfHogwarts("minh.txt"))
```

```python
#W11A4
#Hai chuỗi là anagram của nhau nếu các chữ cái của chuỗi này có thể được sắp xếp để tạo thành chuỗi kia.
#Hãy viết hàm def areAnagrams(path1: str, path2: str) trả về true nếu hai chuỗi đầu vào là anagram của nhau ngược lại false. 
#Hàm không phân biệt chữ hoa chữ thường và không quan tâm đến các dấu câu và dấu cách.
#Đầu vào là 2 tên file trên 2 dòng, mỗi file chứa một chuỗi kí tự.
#Đầu ra: là True nếu 2 chuỗi kí tự là anagram của nhau, False nếu ngược lại

def areAnagrams(path1: str, path2: str):
    file1 = None
    file2 = None
    result = False

    try:
        file1 = open(path1, 'r', encoding='utf-8')
        file2 = open(path2, 'r', encoding='utf-8')
        s1 = file1.read().lower()
        s2 = file2.read().lower()
        clean1 = ""
        clean2 = ""
        for ch in s1:
            if ch.isalpha():
                clean1 += ch
        for ch in s2:
            if ch.isalpha():
                clean2 += ch
        if len(clean1) != len(clean2):
            result = False
        else:
            list1 = list(clean1)
            list2 = list(clean2)
            list1.sort()
            list2.sort()
            result = list1 == list2

    except FileNotFoundError:
        result = False
    finally:
        if file1 is not None:
            file1.close()
        if file2 is not None:
            file2.close()
    return result
print(areAnagrams("minh1.txt", "minh2.txt"))
```

```python
#W11A5
#Viết hàm def count(path: str, character: str), nhận tham số truyền vào là một tên file và một kí tự character, 
#trả về số lần xuất hiện của kí tự character trong chuỗi sentence đọc được từ file.

def count(path: str, character: str):
    file = None
    result = 0

    try:
        file = open(path, 'r', encoding='utf-8')
        sentence = file.read()

        for ch in sentence:
            if ch == character:
                result += 1

    except FileNotFoundError:
        result = 0
    finally:
        if file is not None:
            file.close()

    return result


print(count("minh.txt", "a"))
```

```python
#W11A6
#Giả sử bạn được giao phó trọng trách thống kê này. Bạn được cấp hàng loạt các tệp văn bản lưu lại thơ, văn, 
#cuộc nói chuyện hàng ngày của một ngôn ngữ. Việc đầu tiên bạn cần làm là viết một hàm def countCharacter(path: str, letter: str) 
#nhận tham số đầu vào là đường dẫn đến tệp văn bản và kí tự cần đếm số lần xuất hiện, hàm trả về số lần xuất hiện của kí tự truyền vào. 
#Lưu ý: bạn cần đếm số lần xuất hiện của kí tự mà không phân biệt chữ hoa chữ thường. 
#Vdụ đoạn văn bản sau có 7kí tự e.
#Everyone deserves a second shot.

def countFrequency(path: str, character: str):
    file = None
    result = 0

    try:
        file = open(path, 'r', encoding='utf-8')
        text = file.read().lower()
        character = character.lower()

        for ch in text:
            if ch == character:
                result += 1

    except FileNotFoundError:
        result = 0
    finally:
        if file is not None:
            file.close()

    return result


print(countFrequency("minh.txt", "e"))
```

```python
#W11A8
#Viết chương trình in các từ Fibonacci thứ 0 đến n (n<=20) và ghi ra file output.txt. Công thức của dãy Fibonacci như sau:
#f(0)=a
#f(1)=b
#f(n)=f(n−1)+f(n−2)

def fibonacciWords(path: str):
    file = None
    out = None

    try:
        file = open(path, 'r', encoding='utf-8')

        a = file.readline().strip()
        b = file.readline().strip()
        n = int(file.readline().strip())

        fib = []
        fib.append(a)
        fib.append(b)

        i = 2
        while i <= n:
            fib.append(fib[i - 1] + fib[i - 2])
            i += 1

        out = open("output.txt", "w", encoding="utf-8")

        i = 0
        while i <= n:
            out.write(fib[i] + "\n")
            i += 1

    except FileNotFoundError:
        pass
    finally:
        if file is not None:
            file.close()
        if out is not None:
            out.close()


fibonacciWords("minh.txt")
```

```python
#W11A10
#Viết hàm def getMoney (path: str) -> list[int] để đọc số tiền chứa trong mỗi ngôi nhà từ tệp đầu vào được lưu theo đường dẫn path và 
#trả về mảng chứa các số tiền này 
#Số tiền của mỗi nhà được lưu trong file cách nhau bởi một khoảng trắng và không biết trước số lượng nhà trong khu phố đó. 
#Hàm trả về mảng chứa số tiền của các nhà trong khu phố. 
#Viết hàm def rob(hourses: list[int]) -> int để tính và trả về số tiền lớn nhất mà bạn có thể trộm được từ con phố đó mà không bị phát hiện

def getMoney(path: str):
    file = None
    houses = []

    try:
        file = open(path, 'r', encoding='utf-8')

        content = file.read().strip()
        if content != "":
            parts = content.split()
            for x in parts:
                houses.append(int(x))

    except FileNotFoundError:
        houses = []
    finally:
        if file is not None:
            file.close()

    return houses


def rob(houses: list[int]) -> int:
    if len(houses) == 0:
        return 0

    if len(houses) == 1:
        return houses[0]

    dp = []

    dp.append(houses[0])
    dp.append(max(houses[0], houses[1]))

    i = 2
    while i < len(houses):
        take = dp[i - 2] + houses[i]
        skip = dp[i - 1]

        if take > skip:
            dp.append(take)
        else:
            dp.append(skip)

        i += 1

    return dp[-1]


money = getMoney("minh.txt")
print(rob(money))
```

```python
#W11A11
#Viết chương trình nhận đầu vào là một xâu kí tự chứa đường dẫn đến một tệp. 
#Thử mở tệp này, nếu không thành công in ra "Mission failed". 
#Trong trường hợp mở được, tệp sẽ chứa các số nguyên cách nhau bởi dấu cách hoặc/và dấu xuống dòng. 
#In ra màn hình số lớn nhất và nhỏ nhất trong tệp cách nhau bởi dấu cách.

def findMinMax(path: str):
    file = None
    result = ""

    try:
        file = open(path, 'r', encoding='utf-8')

        content = file.read().strip()
        if content == "":
            result = ""
        else:
            parts = content.split()
            numbers = []

            for x in parts:
                numbers.append(int(x))

            max_val = numbers[0]
            min_val = numbers[0]

            i = 1
            while i < len(numbers):
                if numbers[i] > max_val:
                    max_val = numbers[i]
                if numbers[i] < min_val:
                    min_val = numbers[i]
                i += 1

            result = str(max_val) + " " + str(min_val)

    except FileNotFoundError:
        result = "Mission failed"
    finally:
        if file is not None:
            file.close()

    return result


print(findMinMax("minh.txt"))
```

```python
#W11A12
#Với một dãy số nguyên cho trước, hãy viết hàm def maximumProduct (path: str) -> int: 
#tìm bộ ba số có tích lớn nhất trong tất cả các bộ số và trả về tích lớn nhất đó.
#Hàm nhận đầu vào là đường dẫn đến tệp lưu mảng số nguyên cho trước. Hàm trả về tích lớn nhất có thể có.
#Dòng đầu tiên của tệp chứa kích cỡ n, dòng kế tiếp chứa n số nguyên cách nhau bởi một khoảng trắng.

def maximumProduct(path: str) -> int:
    file = None
    result = 0

    try:
        file = open(path, 'r', encoding='utf-8')

        n_line = file.readline().strip()
        n = int(n_line)

        numbers_line = file.readline().strip()
        parts = numbers_line.split()

        numbers = []
        for x in parts:
            numbers.append(int(x))

        numbers.sort()

        prod1 = numbers[n - 1] * numbers[n - 2] * numbers[n - 3]
        prod2 = numbers[0] * numbers[1] * numbers[n - 1]

        if prod1 > prod2:
            result = prod1
        else:
            result = prod2

    except FileNotFoundError:
        result = 0
    finally:
        if file is not None:
            file.close()

    return result


print(maximumProduct("minh.txt"))
```

```python
#W11A13
#Viết hàm def moveZeroes (fileName: str) -> list thực hiện các công việc sau: 
#Biết dòng đầu tiên của tệp lưu số nguyên n là số số nguyên dương, dòng tiếp theo chứa n số nguyên dương cần lưu vào mảng. 
#Chuyển tất cả các phần tử của mảng có giá trị bằng 0 về cuối dãy, đồng thời, giữ nguyên vị trí tương đối của các phần tử còn lại trong mảng. 
#Hàm trả lại mảng vừa nhận được 

def moveZeroes(fileName: str) -> list:
    file = None
    result = []

    try:
        file = open(fileName, 'r', encoding='utf-8')

        n_line = file.readline().strip()
        n = int(n_line)

        numbers_line = file.readline().strip()
        parts = numbers_line.split()

        numbers = []
        for x in parts:
            numbers.append(int(x))

        non_zero = []
        zero_count = 0

        i = 0
        while i < n:
            if numbers[i] == 0:
                zero_count += 1
            else:
                non_zero.append(numbers[i])
            i += 1

        j = 0
        while j < zero_count:
            non_zero.append(0)
            j += 1

        result = non_zero

    except FileNotFoundError:
        result = []
    finally:
        if file is not None:
            file.close()

    return result


print(moveZeroes("minh.txt"))
```

```python
#W11A14
#Hãy giúp Frank bằng cách viết hàm def averageTime(path: str) , 
#nhận tham số là xâu kí tự lưu đường dẫn tới tệp .csv và trả về thời gian trung bình hoàn thành vòng đua. 
#Lưu ý, tùy tâm trạng Frank sẽ chạy thử xe nhiều vòng hay ít vòng, nên số vòng đua trong các tệp .csv là không cố định. 
#Kết quả lấy đến 2 chữ số phần thập phân.

def averageTime(path: str):
    file = None
    result = 0.0

    try:
        file = open(path, 'r', encoding='utf-8')

        lines = file.readlines()

        total_time = 0.0
        count = 0

        i = 1
        while i < len(lines):
            line = lines[i].strip()
            if line != "":
                parts = line.split(',')
                time_value = float(parts[1])
                total_time += time_value
                count += 1
            i += 1

        if count > 0:
            avg = total_time / count
            result = round(avg, 2)
        else:
            result = 0.0

    except FileNotFoundError:
        result = 0.0
    finally:
        if file is not None:
            file.close()

    return result


print(averageTime("minh.csv"))
```

```python
#W11A15
#Hãy viết hàm def findMovies (path: str) -> list[str] giúp khán giả tìm ra phim với điểm đánh giá lớn hơn 8.0 và có mô tả Không phải là "nhàm chán" ("boring").
#Hàm nhận đầu vào là đường dẫn đến tệp chứa các poster phim. 
#Hàm trả lại các tên những phim thỏa mãn điều kiện trên. 
#Nếu không tìm thấy phim nào phù hợp, trả về thông báo "All are boring!". 
#Kết quả in theo thứ tự xuất hiện trước sau trong file đầu vào

def findMovies(path: str):
    file = None
    result = []

    try:
        file = open(path, 'r', encoding='utf-8')

        lines = file.readlines()

        i = 1
        while i < len(lines):
            line = lines[i].strip()
            if line != "":
                parts = line.split(',')

                movie_name = parts[1].strip()
                description = parts[2].strip().lower()
                rating = float(parts[3].strip())

                if rating > 8.0 and description != "boring":
                    result.append(movie_name)
            i += 1

        if len(result) == 0:
            return "All are boring!"

    except FileNotFoundError:
        return "All are boring!"
    finally:
        if file is not None:
            file.close()

    return result


print(findMovies("minh.csv"))
```

```python
#W11A16
#Viết chương trình, nhập vào đường dẫn đến một tệp. Tệp này chứa thông tin của các hình vuông, hình tròn và hình chữ nhật. 
#In ra chu vi của các hình trong tệp trên từng dòng, theo thứ tự xuất hiện trong tệp. 
#Kết quả được làm tròn đến hai chữ số thập phân sau dấu phẩy.

import math

def printPerimeters(path: str):
    file = None
    result = []

    try:
        file = open(path, 'r', encoding='utf-8')
        lines = file.readlines()

        i = 0
        while i < len(lines):
            line = lines[i].strip()

            if line.startswith("TYPE"):
                shape_type = line.split(":")[1].strip()

                info_line = lines[i + 1].strip()

                if shape_type == "SQUARE":
                    parts = info_line.split(":")
                    side = float(parts[1].strip())
                    perimeter = 4 * side
                    result.append(f"{perimeter:.2f}")

                elif shape_type == "RECTANGLE":
                    parts = info_line.split()
                    width = float(parts[1])
                    height = float(parts[3])
                    perimeter = 2 * (width + height)
                    result.append(f"{perimeter:.2f}")

                elif shape_type == "CIRCLE":
                    parts = info_line.split(":")
                    radius = float(parts[1].strip())
                    perimeter = 2 * math.pi * radius
                    result.append(f"{perimeter:.2f}")

                i += 2
            else:
                i += 1

    except FileNotFoundError:
        result = []
    finally:
        if file is not None:
            file.close()

    return result


perimeters = printPerimeters("minh.txt")
for x in perimeters:
    print(x)
```

```python
#W11A17
#Với một dãy gồm n số nguyên với n>1, giá trị của từng phần tử tại vị trí i được tính lại bằng tích của tất cả các phần tử còn lại trong dãy số nguyên trừ chính bản thân nó - ProductofArrayExceptSelf 
#Viết hàm def productExceptSelf (path: str) -> list[int]: tính và in ra màn hình dãy số nguyên sau khi được tính lại theo qui tắc trên. 
#Hàm nhận đầu vào là đường dẫn path đến tệp chứa dãy số nguyên. Trả lại dãy sau khi biến đổi.

def productExceptSelf(path: str) -> list:
    file = None
    result = []

    try:
        file = open(path, 'r', encoding='utf-8')

        content = file.read().strip()
        if content == "":
            result = []
        else:
            parts = content.split()
            numbers = []

            i = 0
            while i < len(parts):
                numbers.append(int(parts[i]))
                i += 1

            n = len(numbers)
            result = [1] * n

            i = 0
            while i < n:
                product = 1
                j = 0
                while j < n:
                    if i != j:
                        product = product * numbers[j]
                    j += 1
                result[i] = product
                i += 1

    except FileNotFoundError:
        result = []
    finally:
        if file is not None:
            file.close()

    return result


print(productExceptSelf("minh.txt"))
```

```python
#W11A18
#Bạn nhận được một tệp văn bản chứa tọa độ điểm xuất phát, tọa độ mục tiêu và 
#vận tốc của một tên lửa (giả sử tên lửa bay thẳng với vận tốc không đổi). 
#Hãy viết một hàm tính thời gian tên lửa cần để bay tới tiêu diệt mục tiêu của mình: def estimatedTime(path: str). 
#Hàm nhận đầu vào là xâu kí tự chứa đường dẫn tới tệp thiết lập tên lửa và trả về số giây tên lửa hoàn thành hành trình. 
#Đơn vị đo khoảng cách là km và đo tốc độ là km/s.

import math

def estimatedTime(path: str):
    file = None
    result = 0.0

    try:
        file = open(path, 'r', encoding='utf-8')
        lines = file.readlines()

        src_x = 0.0
        src_y = 0.0
        des_x = 0.0
        des_y = 0.0
        velocity = 0.0

        i = 0
        while i < len(lines):
            line = lines[i].strip()

            if line.startswith("SRC"):
                parts = line.split()
                src_x = float(parts[2])
                src_y = float(parts[4])

            elif line.startswith("DES"):
                parts = line.split()
                des_x = float(parts[2])
                des_y = float(parts[4])

            elif line.startswith("VELOCITY"):
                parts = line.split()
                velocity = float(parts[1])

            i += 1

        dx = des_x - src_x
        dy = des_y - src_y

        distance = math.sqrt(dx * dx + dy * dy)

        time = distance / velocity
        result = round(time, 2)

    except FileNotFoundError:
        result = 0.0
    finally:
        if file is not None:
            file.close()

    return result


print(estimatedTime("minh.txt"))
```

```python
#W11A19
#Viết chương trình nhận vào từ bàn phím một chuỗi kí tự là đường dẫn đến một tệp nhị phân. 
#Tệp này chỉ chứa các số thực (kiểu double). 
#In ra các số thực đó (làm tròn đến 5 chữ số thập phân sau dấu phẩy), mỗi số trên một dòng.

import struct

def printDoubles(path: str):
    file = None
    result = []

    try:
        file = open(path, 'rb')

        data = file.read()
        size = 8

        i = 0
        while i + size <= len(data):
            value = struct.unpack('d', data[i:i+size])[0]
            result.append(f"{value:.5f}")
            i += size

    except FileNotFoundError:
        result = []
    finally:
        if file is not None:
            file.close()

    return result


numbers = printDoubles(input())
for x in numbers:
    print(x)
```

```python
#W11A20
#Viết hàm def imageSmoother (path: str) thực hiện công việc trên.
#Hàm nhận đầu vào là đường dẫn đến tệp chứa file ảnh. Hàm thực hiện việc làm mịn ảnh và in ra ảnh sau khi làm mịn.
#Dòng đầu tiên trong tệp chứa kích thước M của ma trận hai chiều. M dòng tiếp theo mỗi dòng chứa M pixel ứng với thông tin của ảnh.

def imageSmoother(path: str):
    file = None
    result = []

    try:
        file = open(path, 'r', encoding='utf-8')

        first_line = file.readline().strip()
        M = int(first_line)

        image = []
        i = 0
        while i < M:
            line = file.readline().strip()
            parts = line.split()
            row = []

            j = 0
            while j < M:
                row.append(int(parts[j]))
                j += 1

            image.append(row)
            i += 1

        new_image = []

        i = 0
        while i < M:
            new_row = []
            j = 0
            while j < M:
                total = 0
                count = 0

                x = i - 1
                while x <= i + 1:
                    y = j - 1
                    while y <= j + 1:
                        if x >= 0 and x < M and y >= 0 and y < M:
                            total += image[x][y]
                            count += 1
                        y += 1
                    x += 1

                new_value = total // count
                new_row.append(new_value)
                j += 1

            new_image.append(new_row)
            i += 1

        result = new_image

    except FileNotFoundError:
        result = []
    finally:
        if file is not None:
            file.close()

    return result


smoothed = imageSmoother("minh.txt")
for row in smoothed:
    print(" ".join(str(x) for x in row))
```
