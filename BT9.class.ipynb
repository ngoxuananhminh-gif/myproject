```python
#W8A1 [Cylinder]
#Xây dựng lớp Cylinder (hình trụ) gồm: bán kính và chiều cao (kiểu số thực) với hai phương thức:tính
#diện tích bề mặt (diện tích toàn phần) và thể tích của hình trụ
#Đầu vào
#Đâu vào từ bàn phím gồm một dòng duy nhất
#chứa bán kính và chiều cao của hình trụ, cách
#nhau bởi một dấu cách.
#Đầu ra
#In ra màn hình diện tích bề mặt và thể tích của
#hình trụ (sử dụng lại những hàm vừa viết, π=3.14).
#(Kết quả làm tròn đến 2 chữ số sau dấu phẩy).

class Cylinder:
    def __init__(self, radius, height):
        self.radius = radius
        self.height = height

    def surface_area(self):
        pi = 3.14
        return 2 * pi * self.radius ** 2 + 2 * pi * self.radius * self.height

    def volume(self):
        pi = 3.14
        return pi * self.radius ** 2 * self.height


# Đọc dữ liệu đầu vào
r, h = map(float, input().split())

# Tạo đối tượng hình trụ
cylinder = Cylinder(r, h)

# In kết quả (làm tròn 2 chữ số sau dấu phẩy)
print(f"{cylinder.surface_area():.2f}")
print(f"{cylinder.volume():.2f}")
```

```python
#W8A2 [Date]
#Xây dựng lớp Date biểu diễn ngày tháng năm với hai phương thức kiểm tra xem một ngày tháng
#nào đó có hợp lệ hay không và phương thức tính ngày kế tiếp của ngày tháng năm nào đó nếu nó là hợp lệ
#Nhập vào một ngày tháng năm, (ngày tháng năm biểu diễn dưới dạng dd/mm/yyyy). Nếu ngày
#tháng năm không tồn tại, in ra dòng chữ INVALID.
#Nếu là hợp lệ in ra ngày kế tiếp của nó.
#Đầu vào
#Đầu vào từ bàn phim chứa dữ liệu ngày, tháng, năm dưới dạng dd/mm/yyyy .
#Đầu ra
#In ra màn hình ngày kế tiếp của ngày vừa nhập dưới dạng dd/mm/yyyy
#Nếu ngày tháng không tồn tại, in ra INVALID 

class Date:
    def __init__ (self, d, m, y):
        self.d = d
        self.m = m
        self.y = y
    def leap_year(self):
        return (self.y % 400 == 0) or (self.y % 4 == 0 and self.y % 100 != 0)
    def valid_date(self):
        if self.y < 1 or self.m < 1 or self.m > 12:
            return False
        days = [31, 29 if self.leap_year() else 28, 31, 30, 31, 30, 31, 31, 30, 31, 30, 31]
        if self.d < 1 or self.d > days[self.m - 1]:
            return False
        return True

    def next_day(self):
        days = [31, 29 if self.leap_year() else 28, 31, 30, 31, 30, 31, 31, 30, 31, 30, 31]
        self.d += 1
        if self.d > days[self.m - 1]:
            self.d = 1
            self.m += 1
            if self.m > 12:
                self.m = 1
                self.y += 1
        return f"{self.d:02d}/{self.m:02d}/{self.y:04d}"

s = input().strip()
d, m, y = map(int, s.split("/"))

date = Date(d, m, y)

if not date.valid_date():
    print("INVALID")
else:
    print(date.next_day())
    
```

```python
#W8A6 Yêu cầu
#Xây dựng lớp Rectangle(w, h) với các phương thức:
#area() trả về diện tích.
#perimeter() trả về chu vi.
#scale(k) phóng to hình chữ nhật theo hệ số k (nhân cả w và h), không trả về gì.
#Input
#w h k (số thực dương)
#Output
#area perimeter sau khi phóng to (làm tròn 2 chữ số thập phân).

class Rectangle:
    def __init__(self, w, h):
        self.w = w
        self.h = h

    def area(self):
        return self.w * self.h

    def perimeter(self):
        return 2 * (self.w + self.h)

    def scale(self, k):
        self.w *= k
        self.h *= k


# ===== MAIN =====
w, h, k = map(float, input().split())

rect = Rectangle(w, h)

# Phóng to hình chữ nhật
rect.scale(k)

# In kết quả
print(f"{rect.area():.2f} {rect.perimeter():.2f}")

        
    
```

```python
#W8A7 Yêu cầu
#Lớp Fraction(num, den) (tử, mẫu nguyên, mẫu ≠ 0) với:
#Rút gọn phân số trong __init__ (quy ước mẫu > 0).
#Các phương thức add/sub/mul/div trả về phân số đã rút gọn.
#__str__() in dạng a/b.
#Input
#a b op c d (tương ứng a/b (op) c/d, op thuộc {+,-,*, /})
#Output
#Phân số kết quả đã rút gọn

import math

class Fraction:
    def __init__(self, num, den):
        if den == 0:
            raise ValueError("Mau so phai khac 0")

        # Quy ước mẫu dương
        if den < 0:
            num = -num
            den = -den

        # Rút gọn phân số
        g = math.gcd(num, den)
        self.num = num // g
        self.den = den // g

    def add(self, other):
        return Fraction(self.num * other.den + other.num * self.den,
                        self.den * other.den)

    def sub(self, other):
        return Fraction(self.num * other.den - other.num * self.den,
                        self.den * other.den)

    def mul(self, other):
        return Fraction(self.num * other.num,
                        self.den * other.den)

    def div(self, other):
        return Fraction(self.num * other.den,
                        self.den * other.num)

    def __str__(self):
        return f"{self.num}/{self.den}"


# ===== MAIN =====
a, b, op, c, d = input().split()
a, b, c, d = int(a), int(b), int(c), int(d)

f1 = Fraction(a, b)
f2 = Fraction(c, d)

if op == "+":
    result = f1.add(f2)
elif op == "-":
    result = f1.sub(f2)
elif op == "*":
    result = f1.mul(f2)
elif op == "/":
    result = f1.div(f2)

print(result)
```

```python
#W8A8 Yêu cầu
#Lớp BankAccount(owner, balance) với:
#deposit(x), withdraw(x); nếu rút quá số dư thì bỏ qua giao dịch.
#__str__() in owner: balance.
#Input
#Dòng 1: owner initial_balance
#Dòng 2: n (số lệnh)
#Tiếp theo n dòng, mỗi dòng: DEPOSIT x hoặc WITHDRAW x.
#Output
#Số dư cuối cùng (in số thực, 2 chữ số).

class BankAccount:
    def __init__(self, owner, balance):
        self.owner = owner
        self.balance = balance

    def deposit(self, x):
        self.balance += x

    def withdraw(self, x):
        if x <= self.balance:
            self.balance -= x
        # nếu rút quá số dư thì bỏ qua

    def __str__(self):
        return f"{self.owner}: {self.balance:.2f}"


# ===== MAIN =====
owner, initial_balance = input().split()
balance = float(initial_balance)

account = BankAccount(owner, balance)

n = int(input())

for _ in range(n):
    cmd, amount = input().split()
    amount = float(amount)

    if cmd == "DEPOSIT":
        account.deposit(amount)
    elif cmd == "WITHDRAW":
        account.withdraw(amount)

# In số dư cuối cùng
print(f"{account.balance:.2f}")
```

```python
#W8A9 Yêu cầu
#Lớp Student(name) với: add_score(subject, score)
#avg() trả về điểm trung bình.
#rank() trả về "Excellent" nếu avg ≥ 8, "Good" nếu 6.5–<8, "Average" nếu 5–<6.5, ngược lại "Poor".
#Input
#Dòng 1: name
#Dòng 2: n
#n dòng tiếp: subject score (score là số thực 0–10)
#Output
#name avg rank (avg 2 chữ số).

class Student:
    def __init__(self):
        self.tablet = []
    
    def add_score(self, subject, score):
        d = {"subject": subject, "score": score}
        self.tablet.append(d)
    
    def avg(self):
        total = 0
        n = len(self.tablet)
        for i in self.tablet:
            total += i["score"]
        result = total/n
        return result
    
    def rank(self):
        if self.avg() >= 8:
            return "Good"
        
        elif 6.5 <= self.avg() < 8:
            return "Average"
        
        elif self.avg() < 6.5:
            return "Poor"

student = Student()
print(input("Nhap ten: "))
n = int(input("Nhap so mon: "))

for i in range (n):
    student.add_score(input("Nhap mon: "), float(input("Nhap diem: ")))

print(f"{student.avg():.2f}")
print(student.rank())
```

```python
#W8A11
# Bạn được cho bốn điểm A, B, C và D trong hệ tọa
# độ Đề-các 3 chiều.
# Nhiệm vụ của bạn là tính và in ra góc giữa hai mặt
# phẳng tạo bởi các điểm:
# mặt phẳng (A, B, C)
# mặt phẳng (B, C, D)
# Góc cần in ra được ký hiệu là PHI, và phải dùng
# độ (degrees), không dùng radian.

import math

class Point:
    def __init__(self, x, y, z):
        self.x = x
        self.y = y
        self.z = z

    def __sub__(self, other):
        return Point(self.x - other.x, self.y - other.y, self.z - other.z)

    def dot(self, other):
        return self.x * other.x + self.y * other.y + self.z * other.z

    def cross(self, other):
        return Point(self.y * other.z - self.z * other.y, self.z * other.x - self.x * other.z, self.x * other.y - self.y * other.x)

    def absolute(self):
        return math.sqrt(self.x**2 + self.y**2 + self.z**2)

def solve():
    points = []

    for _ in range(4):
        line = list(map(float, input().split()))
        points.append(Point(line[0], line[1], line[2]))

    a, b, c, d = points[0], points[1], points[2], points[3]

    ab = b - a
    bc = c - b
    n1 = ab.cross(bc)

    bd = d - b
    n2 = bc.cross(bd) 

    tu_so = abs(n1.dot(n2))
    
    mau_so = n1.absolute() * n2.absolute()

    cos_phi = tu_so / mau_so

    rad = math.acos(cos_phi)
    deg = math.degrees(rad)

    print(f"{deg:.2f}")

if __name__ == "__main__":
    solve()
```

```python
#W8A12
# [Complex]
# Khai báo kiểu dữ liệu biểu diễn số phức \
# (ComplexNumber\) bao gồm \(2\) biến \(realPart\)
# và \(imaginaryPart\) kiểu \(double\) biểu diễn phần
# thực và phần ảo của một số phức.
# Viết hàm khởi tạo ComplexNumber(double
# initRealPart, double initImaginaryPart) để khởi tạo
# các giá trị phần thực và phần ảo của một số phức
# và hàm ComplexNumber multiply(complexNumber
# other) trả về tích của số phức hiện tại và một số
# phức khác.

class ComplexNumber:
    def __init__(self, initRealPart, initImaginaryPart):
        self.realPart = float(initRealPart)
        self.imaginaryPart = float(initImaginaryPart)

    def multiply(self, other):
        new_real = (self.realPart * other.realPart) - (self.imaginaryPart * other.imaginaryPart)
        new_imag = (self.realPart * other.imaginaryPart) + (self.imaginaryPart * other.realPart)
        
        return ComplexNumber(new_real, new_imag)

if __name__ == "__main__":
    input_1 = input().split()
    r1, i1 = float(input_1[0]), float(input_1[1])
    num1 = ComplexNumber(r1, i1)

    input_2 = input().split()
    r2, i2 = float(input_2[0]), float(input_2[1])
    num2 = ComplexNumber(r2, i2)

    result = num1.multiply(num2)

    print(f"{result.realPart} {result.imaginaryPart}")
```

```python
#W8A13
# Viết hàm khởi tạo Triangle(int initEdge1,int
# initEdge2, int initEdge3) để khởi tạo độ dài các
# cạnh của tam giác và hàm double getArea() trả về
# diện tích của tam giác đó (nếu tam giác không
# hợp lệ, trả về \(-1\)).
# Hoàn thiện chương trình dưới đây nhập vào \(3\)
# cạnh của một tam giác và in ra diện tích của tam
# giác đó.

import math

class Triangle:
    def __init__(self, initEdge1, initEdge2, initEdge3):
        self.edge1 = initEdge1
        self.edge2 = initEdge2
        self.edge3 = initEdge3

    def getArea(self):
        a, b, c = self.edge1, self.edge2, self.edge3

        # Kiểm tra tam giác hợp lệ
        if a <= 0 or b <= 0 or c <= 0:
            return -1
        if a + b <= c or a + c <= b or b + c <= a:
            return -1

        # Công thức Heron
        p = (a + b + c) / 2
        area = math.sqrt(p * (p - a) * (p - b) * (p - c))
        return area

edges = list(map(int, input().split()))
triangle = Triangle(edges[0], edges[1], edges[2])

area = triangle.getArea()
if area == -1:
    print("invalid")
else:
    print(f"{area:.2f}")
```
