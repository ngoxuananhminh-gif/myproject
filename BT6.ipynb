```python
#1 Nhập vào một dòng gồm các số nguyên. Hãy loại bỏ phần tử trùng nhưng
#giữ nguyên thứ tự xuất hiện đầu tiên, rồi in ra list kết quả.
#Yêu cầu in ra đúng dạng list Python, ví dụ [1, 2, 3]

nums = list(map(int,input("Nhap day so nguyen: ").split()))
result = []
seen = {}

for x in nums:
    if x not in seen:
        result.append(x)
        seen[x] = True


print(result)

```

```python
#2 Nhập một dòng số nguyên. In ra list tổng tích lũy (phần tử i là tổng từ đầu
#tới vị trí i), giữ nguyên thứ tự

nums = list(map(int,input("Nhap day so nguyen: ").split()))

result = []
tong = 0

for x in nums:
    tong += x
    result.append(tong)

print(result)
```

```python
#3 Nhập một tuple số nguyên (một dòng) và số nguyên k (dòng kế). In ra tuple
#sau khi xoay trái k vị trí. (Giữ nguyên thứ tự tương đối sau xoay; không sắp
#xếp.)
                               

t = list(map(int,input().split()))
k = int(input())

n = len(t)
k %= n
lst = list(t)

for _ in range(k):
    first = lst[0]
    lst = lst[1:] + [first]

result = tuple(lst)

print(result)
```

```python
#4 Nhập một dòng gồm các cặp key:value cách nhau bởi dấu cách. Tạo
#dictionary ánh xạ mỗi key -> list các value theo thứ tự xuất hiện. Nếu key
#lặp lại, thêm vào list của key đó

pairs = input().split()

d = {}   

for p in pairs:
    key, value = p.split(":")   
    value = int(value)          

    if key not in d:            
        d[key] = [value]        
    else:
        d[key].append(value)    

print(d)
```

```python
#5 Nhập một list số nguyên, In ra dictionary {'positives': ..., 'negatives': ...,
#'zeros': ...} (đếm số >0, <0, =0)

nums = list(map(int, input().split()))

pos = neg = zero = 0

for x in nums:
    if x > 0:
        pos += 1
    elif x < 0:
        neg += 1
    else:
        zero += 1
result = {
    'positives': pos,
    'negatives': neg,
    'zeros': zero
}

print(result)
```

```python
#6 Viết chương trình nhập vào một danh sách các số nguyên, sau đó in ra tổng
#các số trong list đó.


nums = list(map(int, input().split()))
total = sum(nums)

print(total)
```

```python
#7 Viết chương trình nhập vào một tuple, in ra phần tử đầu, phần tử cuối, và
#tuple đảo ngược


t = tuple(map(int, input().split()))

first = t[0] if len(t) > 0 else None
last = t[-1] if len(t) > 0 else None
reversed_t = t[::-1]

print("Phần tử đầu:", first)
print("Phần tử cuối:", last)
print("Tuple đảo ngược:", reversed_t)
```

```python
#8 Viết chương trình nhập vào một list các chuỗi, sau đó đếm số lần xuất hiện
#của từng chuỗi và lưu vào dictionary.

words = input().split()

count_dict = {}

for word in words:
    if word in count_dict:
        count_dict[word] += 1
    else:
        count_dict[word] = 1

print(count_dict)
```

```python
#9 Viết chương trình nhập vào hai dictionary. Sau đó gộp lại thành một
#dictionary mới trong đó nếu một key xuất hiện trong cả hai dict, giá trị sẽ
#được cộng lại, nếu chỉ có trong một dict thì giữ nguyên. In ra dictionary mới
#theo thứ tự từ điển tăng dần của key. 

dict1 = {}
for item in input("Nhập dict1 (key:value cách nhau bằng khoảng trắng): ").split():
    key, value = item.split(":")
    dict1[key] = int(value)
dict2 = {}
for item in input("Nhập dict2 (key:value cách nhau bằng khoảng trắng): ").split():
    key, value = item.split(":")
    dict2[key] = int(value)
merged = dict1.copy()  
for key, value in dict2.items():
    if key in merged:
        merged[key] += value
    else:
        merged[key] = value

merged_sorted = dict(sorted(merged.items()))

print(merged_sorted)
```

```python
#10 Viết chương trình nhập vào một danh sách các số nguyên, và số nguyên
#dương k. In ra list gồm các cặp số khác nhau có tổng bằng k. Mỗi cặp là
#một tuple, hai số trong tuple có thể giống nhau, thứ tự xuất hiện trong danh
#sách của số thứ nhất nhỏ hơn thứ tự của số thứ hai. Các tuple trong list được
#sắp xếp theo thứ tự tăng dần của phần tử đầu tiên trong tuple.

nums = list(map(int, input("Nhập danh sách số nguyên: ").split()))

k = int(input("Nhập số nguyên dương k: "))

pairs = []

n = len(nums)
for i in range(n):
    for j in range(i+1, n):  
        if nums[i] + nums[j] == k:
            pairs.append((nums[i], nums[j]))

pairs.sort(key=lambda x: x[0])

print(pairs)
```

```python
#11 [Tách số chẵn và lẻ từ tuple]
#Nhập vào một tuple chứa các số nguyên. Hãy tạo ra hai tuple mới: một chứa
#tất cả các số chẵn và một chứa tất cả các số lẻ từ tuple ban đầu, giữ nguyên
#thứ tự tương đối. In ra hai tuple này trên hai dòng riêng biệt.

t = tuple(map(int, input("Nhập tuple: ").split()))

even = tuple(x for x in t if x % 2 == 0)
odd = tuple(x for x in t if x % 2 != 0)

print(even)
print(odd)
```

```python
#12 [Tìm phần tử xuất hiện nhiều nhất]
#Nhập vào một list các số nguyên. Hãy tìm số xuất hiện nhiều lần nhất trong
#list. Nếu có nhiều số cùng có số lần xuất hiện nhiều nhất, hãy in ra số nhỏ
#nhất trong số đó.

nums = list(map(int, input("Nhập list số nguyên: ").split()))

count_dict = {}
for num in nums:
    if num in count_dict:
        count_dict[num] += 1
    else:
        count_dict[num] = 1

max_count = max(count_dict.values())
candidates = [num for num, count in count_dict.items() if count == max_count]

print(min(candidates))
```

```python
#13 [Đảo ngược key và value của dictionary]
#Nhập vào một dictionary có value là duy nhất (không trùng nhau). Hãy tạo
#một dictionary mới bằng cách đảo ngược key và value của dictionary ban
#đầu. In ra dictionary mới.

d = {}
for item in input("Nhập dictionary (key:value cách nhau bằng khoảng trắng): ").split():
    key, value = item.split(":")
    d[key] = value  
reversed_dict = {v: k for k, v in d.items()}

print(reversed_dict)
```

```python
#14 [Giao của hai danh sách]
#Nhập vào hai dòng, mỗi dòng là một danh sách các số nguyên. Hãy tìm các
#phần tử chung (giao) của hai danh sách này và in ra một list mới chứa các
#phần tử chung đó, không trùng lặp.

list1 = list(map(int, input("Nhập danh sách 1: ").split()))
list2 = list(map(int, input("Nhập danh sách 2: ").split()))
intersection = list(set(list1) & set(list2))

print(intersection)
```

```python
#15 [Lọc dictionary theo giá trị]
#Nhập vào một dictionary có key là chuỗi và value là số nguyên, và một số
#nguyên k ở dòng tiếp theo. Hãy tạo một dictionary mới chỉ chứa các cặp
#key-value từ dictionary ban đầu mà có value lớn hơn k. In ra dictionary
#mới.

d = {}
for item in input("Nhập dictionary (key:value cách nhau bằng khoảng trắng): ").split():
    key, value = item.split(":")
    d[key] = int(value)
k = int(input("Nhập số nguyên k: "))
filtered = {key: value for key, value in d.items() if value > k}

print(filtered)
```
