```python
#Bai 5
import math
a,b,c = map(float, input().split())
if a == 0:
    if b == 0:
        if c == 0:
            print("PT vo so nghiem")
        else:
            print("PT vo nghiem")
    else:
        x = -c / b
        print(f"PT co mot nghiem: x = {x:.2f}")
else:
    delta = b**2 - 4 * a * c

    if delta < 0:
        print("PT vo nghiem")
    elif delta == 0:
        x = -b / (2 * a)
        print(f"PT co nghiem kep: x = {x:.2f}")
    else:
        x1 = (-b + math.sqrt(delta)) / (2 * a)
        x2 = (-b - math.sqrt(delta)) / (2 * a)
        print("PT co hai nghiem phan biet:")
        print(f"x1 = {x1:.2f}")
        print(f"x2 = {x2:.2f}")
```
