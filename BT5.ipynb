```python
#1 Viết hàm truyền vào hai số nguyên a, b. Trả về số lớn hơn trong hai số. def Max_Of_Two(a, b) -> int:
def Max_Of_Two(a: int, b: int) -> int:
    if a > b:
        return a
    else:
        return b
print(Max_Of_Two(9, 10))   
```

```python
#3 Viết hàm kiểm tra một số có phải là số nguyên tố hay không? 
import math
def check_prime(n):
    if n > 1:
        for i in range(2, int(math.sqrt(n)) + 1):
            if n % i == 0:
                return False
        return True
    return False
    
print(check_prime(10))
```

```python
#4 VIết hàm kiểm tra một số có phải số hoàn hảo hay không, xử lý trên dữ liệu kiểu số nguyên.
def perfect_number(n):
    if n > 0:
        tong = 0
        for i in range(1, n):
            if n % i == 0:
                tong += i
                if tong == n:
                    return True
    return False
print(perfect_number(28))
```

```python
#5 Viết hàm đưa vào một list số nguyên và một số nguyên dương k. Hãy tìm và trả về vị trí của phần tử đầu tiên có giá trị k trong list số nguyên, nếu không có thì trả về -1 
from typing import List

def find_first_index(lst: List[int], k: int) -> int:
    for i in range(len(lst)):
        if lst[i] == k:
            return i  
    return -1 

print(find_first_index([1, 2, 3, 4, 5], 10))
```

```python
#6 Viết hàm tính giai thừa. Nhập vào số nguyên n, trả về giai thừa của n (n!)
def factorial(n: int) -> int:
    """
    Hàm tính giai thừa n! của số nguyên n >= 0.
    """
    if n < 0:
        return None  # giai thừa chỉ định nghĩa cho n >= 0

    result = 1
    for i in range(1, n + 1):
        result *= i  # nhân liên tiếp từ 1 đến n
    return result
```

```python
#8 Khoảng cách Hamming giữa hai số nguyên là số lượng vị trí khác nhau giữa hai dãy bits
#tương ứng của chúng.
def hamming_distance(x, y):
    count = 0
    while x > 0 or y > 0:     
        bit_x = x % 2
        bit_y = y % 2
        if bit_x != bit_y:
            count += 1
        x //= 2
        y //= 2
    return count

print(hamming_distance(7, 10))
   
```

```python
#9 Viết hàm trả về tổng các chữ số của số nguyên dương n 
def sum_digits(n: int) -> int:
    
    if n < 0:
        return None  

    total = 0
    while n > 0:
        total += n % 10  
        n //= 10         
    return total

print(sum_digits(100))
```

```python
#10 Hai từ được gọi là đẳng cấu nếu các kí tự trong một từ có thể được ánh xạ để được từ còn
#lại. Ánh xạ một kí tự nghĩa là thay đổi toàn bộ các kí tự đó trong từ bởi một kí tự khác. Thứ
#tự của các kí tự được giữ không đổi. Hai kí tự khác nhau không thể ánh xạ đến cùng một kí
#tự, nhưng một kí tự có thể ánh xạ thành chính nó. Ví dụ, từ "abca" và "zbxz" là hai từ đẳng
#cấu vì ta có thể ánh xạ 'a' thành 'z', 'b' thành 'b' và 'c' thành 'x'.
#Viết hàm trả về \(true\) nếu hai từ \(a\) và \(b\) là đẳng cấu, ngược lại \(false\). Bạn có thể
#giả sử hai từ đầu vào có cùng độ dài.  

def isIsomorphic(a, b):
    n = len(a)

    i = 0
    while i < n:
        j = 0
        while j < n:
            if (a[i] == a[j]) != (b[i] == b[j]):
                return False
            j += 1
        i += 1
    
    return True

print(isIsomorphic("abcd", "xyzd"))
```

```python
#10 Hai từ được gọi là đẳng cấu nếu các kí tự trong một từ có thể được ánh xạ để được từ còn
#lại. Ánh xạ một kí tự nghĩa là thay đổi toàn bộ các kí tự đó trong từ bởi một kí tự khác. Thứ
#tự của các kí tự được giữ không đổi. Hai kí tự khác nhau không thể ánh xạ đến cùng một kí
#tự, nhưng một kí tự có thể ánh xạ thành chính nó. Ví dụ, từ "abca" và "zbxz" là hai từ đẳng
#cấu vì ta có thể ánh xạ 'a' thành 'z', 'b' thành 'b' và 'c' thành 'x'.
#Viết hàm trả về \(true\) nếu hai từ \(a\) và \(b\) là đẳng cấu, ngược lại \(false\). Bạn có thể
#giả sử hai từ đầu vào có cùng độ dài.  
def is_isomorphic(a: str, b: str) -> bool:
    
    map_a_to_b = {}
    map_b_to_a = {}

    for char_a, char_b in zip(a, b):
        
        if char_a in map_a_to_b:
            if map_a_to_b[char_a] != char_b:
                return False
        else:
            map_a_to_b[char_a] = char_b

       
        if char_b in map_b_to_a:
            if map_b_to_a[char_b] != char_a:
                return False
        else:
            map_b_to_a[char_b] = char_a

    return True

print(is_isomorphic("abcd", "xyzd"))
```
