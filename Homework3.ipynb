```python
#4 Viết chương trình in ra hình tam giác cân có độ cao h được nhập từ bàn phím 
h = int(input("Nhập vào độ cao h: "))
for i in range (1, h + 1):
    if i == 1:
        print( " " * (h - i) + "*")
    elif i == h:
        print( "*" * (i*2-1) )
    else:
        print( " " * (h-i) + "*" + " " * (2*i-3) + "*" )
```

```python
#9 Nhập vào số nguyên dương a, kiểm tra xem a có phải là số chính phương hay không? Số chính phương là số có căn bậc hai là một số nguyên VD: 1, 4, 9, 16, 25…. 
a = int(input("Nhập vào so nguyen duong a: "))
found = False
for i in range(1, a + 1):
    if i * i == a:
        found = True
        break
if found:
    print(f"{a} là số chính phương")
else:
    print(f"{a} không phải là số chính phương")

```

```python
#Bài 12: Nhập số nguyên dương n, Cho S(k) = 1 + 2 + 3 + … + k. Tìm k sao cho S(k) lớn nhất nhưng nhỏ hơn n 
while True:
    n = int(input("Nhập một số nguyên dương: "))
    if n > 0:
        break
S = 0        
k = 1

while True:
    S += k
    k += 1
    if S >= n:
        k -= 1
        break
print("k can tim sao cho maxS(k) < n la:", k)       
```

```python
#Bài 13: Nhập vào số nguyên dương A, tìm n nhỏ nhất sao cho
#1 + 1/2 + 1/3 + 1/4 + ... + 1/n > A 

while True:
    A = int(input("Nhập một số nguyên dương: "))
    if A > 0:
        break

S = 0
n = 0
while S <= A:
    n += 1
    S += (1/n)
      
print("So can tim n la:", n)
    
```

```python
#Bài 14: Nhập vào một dãy số nguyên, dừng nhập khi người dùng nhập -1. Sau khi nhập
#xong, in số lớn nhất, số nhỏ nhất trong những số vừa nhập. KHÔNG ĐƯỢC dùng mảng 
max_n = None
min_n = None

while True:
    n = int(input("Nhap so nguyen: "))
    if n == -1:
        break
    
    if max_n == None:
        max_n = n
        min_n = n
    
    else:
        if n > max_n:
            max_n = n
        if n < min_n:
            min_n = n

print(max_n, "la so lon nhat cua n")
print(min_n, "la so nho nhat cua n")
```

```python
#Bài 19: Dãy số fibonacci là dãy số được định nghĩa như sau: 1, 1, 2, 3, 5, 8, 13,... với số
#kế tiếp sẽ bằng tổng hai số trước đó. Nhập vào số nguyên dương A, hãy tìm số trong
#dãy số fibonacci lớn nhất nhưng không vượt quá A 

A = int(input("Nhập số nguyên dương A: "))

f1 = 1
f2 = 1
n = 2   

while True:
    f_next = f1 + f2
    n += 1
    if f_next > A:
        n -= 1
        break
    f1 = f2
    f2 = f_next

print(n)

```
