```python
#8 Nhập số nguyên dương n. Lặp lại phép toán n = n + reverse(n) cho đến khi n là số đối xứng (palindrome). In ra số bước và giá trị palindrome thu được.
def reverse(n):
    rev = 0
    while n > 0:
        rev = rev * 10 + n % 10  
        n = n // 10              
    return rev

def palindrome(n):
    return n == reverse(n)

def process(n):
    step = 0
    while not palindrome(n):
        n += reverse(n)
        step += 1
    return step, n
n = int(input("Nhập số nguyên dương n: "))

step, n = process(n)
print("So buoc:", step)
print("Gia tri:", n)

```

```python
#9 [PerfectSquaresDistinctDigits] 
#Nhập vào số nguyên dương n. In ra tất cả các số chính phương ≤ n có các chữ số đôi một khác nhau 
#(in theo thứ tự tăng dần, cùng hàng, cách nhau bởi một dấu cách). Nếu không có, in ra (no number). 

n = int(input("Nhập số nguyên dương n: "))
result = []
for i in range (1, n+1):
    if i * i <= n:
        scp = i * i
        chuso = str(scp)
        if len(set(chuso)) == len(chuso):
            result.append(chuso)
if result:
    print(" ".join(result))
else:
    print("No number")
```

```python
#10 "[LongestCollatzUpToN] 
#Nhập vào số nguyên dương n. Tìm số x trong [1, n] có độ dài dãy Collatz lớn nhất 
#và in ra x cùng độ dài dãy tương ứng (x L). Nếu có nhiều số cùng độ dài, chọn số 
#nhỏ nhất." 

n = int(input())
max_len = 0        
best_x = 1         

x = 1
while x <= n:
    temp = x
    length = 1     
    while temp != 1:
        if temp % 2 == 0:
            temp //= 2
        else:
            temp = 3 * temp + 1
        length += 1
    if length > max_len:
        max_len = length
        best_x = x
    x += 1

print(best_x, max_len)
```
